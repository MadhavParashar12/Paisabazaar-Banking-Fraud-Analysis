# **Project Summary -**

This project focuses on analysing customer financial and credit-related data to understand the factors associated with **credit risk and credit scores**, with the broader objective of helping Paisabazaar make better, more personalized and responsible financial recommendations.

The analysis began with **data understanding and data quality checks**. The dataset contains **100,000 records** with numerical, categorical and textual variables. We examined missing values, semantic missing values such as *“No Data”, “Not Specified” and “NM”*, duplicate records, data types and descriptive statistics. No duplicate records were found. The data was then prepared for analysis by standardizing column names, removing unnecessary spaces from text fields and checking appropriate data types.

The **Univariate Analysis** helped us understand the distribution of individual variables. Important observations included considerable variation in income, outstanding debt, EMI, credit history age and repayment-related variables. Credit Mix and Credit Score showed three main categories, while Payment Behaviour had six distinct categories.

The **Bivariate Analysis** focused on relationships with Credit Score. Delayed payments showed a clear association with poorer credit scores. Outstanding debt, number of loans and interest rate also showed meaningful differences across credit-score groups. Income showed more overlap between Good, Standard and Poor customers, suggesting that income alone is not enough to identify credit risk. Payment Behaviour and Credit Mix were particularly useful categorical indicators. Customers with a Bad Credit Mix consistently had a much higher proportion of Poor credit scores.

The **Multivariate Analysis** provided deeper insights by combining multiple factors. Higher outstanding debt together with frequent delayed payments was associated with a greater concentration of Poor credit scores. The combination of income, outstanding debt and EMI also highlighted potentially financially stressed customers, particularly those with lower or moderate income and higher debt and repayment burdens. The analysis of Payment Behaviour and Credit Mix identified specific high-risk combinations, with some Bad Credit Mix segments having more than 60% Poor-credit customers.

The **Correlation Heatmap** showed strong relationships among several numerical variables. Annual Income and Monthly Inhand Salary were almost perfectly correlated, indicating redundancy. Income was also strongly related to monthly investments, while loans, debt, inquiries and interest rate showed several moderate-to-strong relationships. Credit utilization, however, had relatively weak linear relationships with most numerical variables. The Pair Plot further supported the importance of delayed payments and credit history while showing substantial overlap for income and credit utilization across credit-score groups.

Overall, the EDA indicates that **credit risk is multidimensional**. No single variable can reliably explain a customer's credit score. The strongest practical signals came from **repayment behaviour, Credit Mix, outstanding debt, EMI burden and credit history**.

Based on these findings, Paisabazaar should adopt a **risk-based customer segmentation approach**. Lower-risk customers can receive suitable credit-product recommendations, while customers showing high debt, frequent delays and poor credit characteristics should receive stronger affordability checks and more responsible recommendations such as credit-building or debt-management solutions.

The overall objective should be to move from simply identifying **“who can be offered credit”** to understanding **“which financial product is appropriate for each customer's financial situation and repayment capacity.”**

## 3. ***Data Wrangling***

### Data Wrangling Code

```python
# Write your code to make your dataset analysis ready.

# create a wrorking copy so that the original dataset remains unchanged.
df = df.copy()
```

```python
# Standardize the column names
df.columns = (
    df.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)
```

```python
# Clean string columns without changing numerical values.
object_columns = df.select_dtypes(include="object").columns

for col in object_columns:
    df[col] = df[col].str.strip()
```

```python
# Check data types after cleaning
print("\nData types after cleaning:")
print(df.dtypes)
```

```python
for col in object_columns:
    converted_col = pd.to_numeric(df[col], errors="coerce")

    # Convert only if all non-null values can be interpreted as numbers.
    if converted_col.notna().sum() == df[col].notna().sum():
        df[col] = converted_col
```

```python
print("\nFinal dataset shape:", df.shape)

print("\nFinal duplicate count:", df.duplicated().sum())

print("\nFinal missing value count:", df.isnull().sum().sum())

print("\nFinal data types:")
print(df.dtypes)
```
